## SKtime test

```python
from sktime.classification.deep_learning import ResNetClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import f1_score

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
fold_results = []
fold_scores = []

for fold_num, (train_idx, test_idx) in enumerate(skf.split(X_torque_angle, y), 1):
    X_train_fold = X_torque_angle[train_idx]
    X_test_fold = X_torque_angle[test_idx]
    y_train_fold = y[train_idx]
    y_test_fold = y[test_idx]
    usage_test_fold = usage[test_idx]
    
    scaler = StandardScaler()
    X_train_flat = X_train_fold.reshape(X_train_fold.shape[0], -1)
    X_train_scaled = scaler.fit_transform(X_train_flat).reshape(X_train_fold.shape)
    
    X_test_flat = X_test_fold.reshape(X_test_fold.shape[0], -1)
    X_test_scaled = scaler.transform(X_test_flat).reshape(X_test_fold.shape)
    
    resnet = ResNetClassifier(n_epochs=30, batch_size=32, verbose=0, random_state=42)
    resnet.fit(X_train_scaled, y_train_fold)
    
    y_pred_fold = np.asarray(resnet.predict(X_test_scaled)).reshape(-1)
    
    fold_results.append({
        "True_label": y_test_fold,
        "Pred_label": y_pred_fold,
        "workpiece_usage": usage_test_fold,
        "fold": fold_num
    })
    
    f1 = f1_score(y_test_fold, y_pred_fold, average="macro")
    fold_scores.append(f1)
    
    print(f"Fold {fold_num}: F1 = {f1:.4f}")

print(f"ResNet Mean F1: {np.mean(fold_scores):.4f} ± {np.std(fold_scores):.4f}")

resnet_binary_s1st = pred_filter(fold_results, label="binary", usage1st=True)
resnet_binary_sall = pred_filter(fold_results, label="binary", usage1st=False)
resnet_single_s1st = pred_filter(fold_results, label="single", usage1st=True)
resnet_single_sall = pred_filter(fold_results, label="single", usage1st=False)
resnet_grouped_s1st = pred_filter(fold_results, label="grouped", usage1st=True)
resnet_grouped_sall = pred_filter(fold_results, label="grouped", usage1st=False)

resnet_cv_results = pd.concat([resnet_binary_s1st, resnet_binary_sall, resnet_single_s1st, resnet_single_sall, resnet_grouped_s1st, resnet_grouped_sall], ignore_index=True)
resnet_cv_results.insert(0, "Labelstrategie",["Binary_s1st", "Binary_all", "Single_s1st", "Single_all", "Grouped_s1st", "Grouped_all"])
resnet_cv_results["Model"] = "ResNet_CV"
resnet_cv_results.to_csv('ResNet_CV_results_X_torque_angle.csv', index=False)
resnet_cv_results
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import f1_score
from sklearn.preprocessing import LabelEncoder

x=np.asarray(x_tsfresh)
y=np.asarray(y_tsfresh)
le = LabelEncoder()
y = le.fit_transform(y)
usage=np.asarray(usage)
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
mlp_fold_results=[]
fold_scores = []
for fold_num, (train_idx, test_idx) in enumerate(skf.split(x,y), 1):
    X_train, X_test = x[train_idx], x[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]
    usage_test = usage[test_idx]
    scaler = StandardScaler()   
    X_train = scaler.fit_transform(X_train)
    X_test = scaler.transform(X_test)
    
    clf = MLPClassifier(hidden_layer_sizes=(256,), activation='relu',
                        solver='adam', max_iter=300, early_stopping=True,
                        random_state=42)
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    mlp_fold_results.append({
        "True_label": y_test,
        "Pred_label": y_pred,
        "workpiece_usage": usage_test,
        "fold": fold_num
    })
    f1 = f1_score(y_test,y_pred,average="macro")
    fold_scores.append(f1)
    print(f"Fold {fold_num} F1-score {f1:.4f}")
print(f"MLP Mean F1: {np.mean(fold_scores):.4f} +- {np.std(fold_scores):.4f}")
df_results = evaluate_folds(mlp_fold_results)
df_results.to_csv('mlp_tsfresh_results.csv', index=False)
df_results
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import f1_score
from sklearn.preprocessing import LabelEncoder
from benchmark_modular.metrics import evaluate_folds
le = LabelEncoder()
y = le.fit_transform(y)
usage=np.asarray(usage)
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
mlp_X_fold_results=[]
fold_scores = []
for fold_num, (train_idx, test_idx) in enumerate(skf.split(X_torque_angle,y), 1): #torque angle
    X_train, X_test = X_torque_angle[train_idx], X_torque_angle[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]
    usage_test = usage[test_idx]
    
    X_train= X_train.reshape(X_train.shape[0], -1)
    X_test = X_test.reshape(X_test.shape[0], -1)
    scaler = StandardScaler()   
    X_train = scaler.fit_transform(X_train) #Found array with dim 3, while dim <= 2 is required by StandardScaler. Erst in 2d umwandeln, dann skalieren, dann wieder in 3d umwandeln
    X_test = scaler.transform(X_test)
    
    clf = MLPClassifier(hidden_layer_sizes=(256,), activation='relu',
                        solver='adam', max_iter=300, early_stopping=True,
                        random_state=42)
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    mlp_X_fold_results.append({
        "True_label": y_test,
        "Pred_label": y_pred,
        "workpiece_usage": usage_test,
        "fold": fold_num
    })
    f1 = f1_score(y_test,y_pred,average="macro")
    fold_scores.append(f1)
    print(f"Fold {fold_num} F1-score {f1:.4f}")
print(f"MLP Mean F1: {np.mean(fold_scores):.4f} +- {np.std(fold_scores):.4f}")
df_results = evaluate_folds(mlp_X_fold_results)
df_results.to_csv('mlp_torque-torque_angle_results.csv', index=False)
df_results
```

## PIML
drehmoment formel screw driving M = k * d * f  
M = Torque (vorhanden)  in Nm
k = Reibungskoeffizient   (kommt später als zusätzliche feature hins training)
d = Durchmesser Schraube (vorhanden)  DELTA PT® 40x12 4.0mm durchmesser x 12mm länge  
F = Vorspannkraft (nicht wirklich vorhanden - oder?)  - konstanten vordefinierten wert (siehe handbuch schraube)



Latent Pinn: k läuft über einen ganzen schraubvorgang (class_value=oiled) und F ändert sich mit der zeitreihe (800) und wird irgendwie durch den winkel definiert. Wenn ich heruasfinde welche werte dabei "realistisch sind", kann man reale boundaries setzen, dadurch lernt das modell welchen anteil, welche variable am torque value hat hat. Somit M = k * d * F bekannt. Mehrfachabrieb durch work_cycles bei k berücksichtigen k(work_cycle) abhängigkeit bei Ermittlung von k. Der loss umfasst dann mehrere constraints https://arxiv.org/pdf/2305.07671  
Symbolic Regression: https://www.iese.fraunhofer.de/blog/symbolic-regression-as-a-path-to-physical-ai/
